```python
# Gets the class corpuses
vio_class_corpuses = data_utils.get_class_corpuses(df_violence, "VIO")

# Tokenizes the class corpuses
vio_true_tokens, vio_true_sentences = data_utils.prepare_text_for_frequency_distribution_analysis(
    vio_class_corpuses["VIO_true"]
)
vio_false_tokens, vio_false_sentences = data_utils.prepare_text_for_frequency_distribution_analysis(
    vio_class_corpuses["VIO_false"]
)

# Creates the frequency distribution classes
vio_true_freq_dist = FreqDist(vio_true_tokens)
vio_false_freq_dist = FreqDist(vio_false_tokens)

# Creates a dictionary with the frequency distribution of the classes
freq_dist_dict_unfiltered = {"VIO_true": vio_true_freq_dist, "VIO_false": vio_false_freq_dist}

# Plots the frequency distributions as a bar plot
vio_freq_dist_fig, vio_freq_dist_ax = data_utils.create_frequency_distribution_plot(
    freq_dist_dict_unfiltered,
    title="VIO Frequency Word Distribution per Class (unfiltered)", 
    n_words=30,
    figsize=(16, 10),
    column_names=["Word", "Frequency"],
    show_grid=True
)
```
