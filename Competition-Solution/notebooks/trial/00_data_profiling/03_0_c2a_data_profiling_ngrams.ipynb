<p><b>C2A n-gram Frequency Distribution (unfiltered)</b></p>
<p><i><li>Bar Plot</li><i><p>

```python
# Gets the class corpuses
c2a_class_corpuses = data_utils.get_class_corpuses(df_call2action, "C2A")

# Tokenizes the class corpuses
c2a_true_tokens, c2a_true_sentences = data_utils.prepare_text_for_frequency_distribution_analysis(
    c2a_class_corpuses["C2A_true"]
)
c2a_false_tokens, c2a_false_sentences = data_utils.prepare_text_for_frequency_distribution_analysis(
    c2a_class_corpuses["C2A_false"]
)

# Creates the n-gram frequency distributions for each class
c2a_true_bigram_freq = data_utils.generate_ngrams_from_tokens(c2a_true_tokens, n=2)
c2a_false_bigram_freq = data_utils.generate_ngrams_from_tokens(c2a_false_tokens, n=2)
c2a_true_trigram_freq = data_utils.generate_ngrams_from_tokens(c2a_true_tokens, n=3)
c2a_false_trigram_freq = data_utils.generate_ngrams_from_tokens(c2a_false_tokens, n=3)

# Creates a dictionary with the frequency distributions
ngram_freq_dict_unfiltered = {"C2A_true_bigram": c2a_true_bigram_freq, "C2A_false_bigram": c2a_false_bigram_freq, 
                              "C2A_true_trigram": c2a_true_trigram_freq, "C2A_false_trigram": c2a_false_trigram_freq}

# Plots the n-gram distributions as a bar plot
c2a_bigram_fig, c2a_bigram_ax = data_utils.create_ngram_frequency_plot(
    ngram_freq_dict_unfiltered,
    title="C2A n-gram Distribution per Class (unfiltered)", 
    n_items=20,
    figsize=(16, 10),
    show_grid=True
)
```
