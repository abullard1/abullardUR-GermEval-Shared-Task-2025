<u><p>Summary & Implication of Tweet Lengths</p></u>

*   *Observation:* Tweets in the trial datasets have a median character length between 77-91 characters, but the distribution is heavily right-skewed due to the presence of a few very long texts (up to ~3000 characters). There are also some very short (1-2 character) tweets.
*   *Implication:*
    *   **Outliers:** Outliers such as very short and very long tweets need to be qualitatively examined. This includes checking if very short tweets are just "noise" and if very long tweets are valid concatenated posts or merely artifacts of some sort. The actions taken will depend on the technical approach taken to classify the tweets. (e.g. Transformer-based models, LSTM, RNNs, etc.)
    *   **Model Input Length (`max_length`):** The presence of very long texts means that a standard `max_length` for many encoder-based models (e.g., 128, 256, or 512) will lead to truncation for these outliers. This is probably acceptable, as most of our tweets fall well within these limits (e.g. 75% are under ~190 characters). The mean, median and standard deviation tweet length values that we have found out for the subtasks are going to aid in choosing a maximum input length that is both performant as well as captures the majority of the relevant data.
