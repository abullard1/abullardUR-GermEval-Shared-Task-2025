##### <b>Overlap between the subtask datasets</b>

```python
datasets_to_compare = [
    ("Call2Action", df_call2action), 
    ("Violence", df_violence), 
    ("DBO", df_dbo)
]

# Calculates the raw overlaps
overlaps = data_utils.calculate_dataset_overlaps(
    datasets=datasets_to_compare,
    column="description",
    show_details=False
)

# Displays metrics
percentage_results = data_utils.display_dataset_overlap_percentages(
    overlaps=overlaps,
    datasets=datasets_to_compare,
    column="description",
    show_pairwise=True
)
```

<u><p>Interpretation:</p></u>

The analysis of overlapping tweet descriptions across the three subtask datasets (Call2Action, Violence, and DBO) reveals the following:

*   **Overall Corpus Overlap:**
    *   Out of a total of **2868 unique descriptions** across all three datasets combined, **264 descriptions (9.21%)** appear in *at least two* of these datasets.
    *   This indicates a moderate level of shared content when considering the entire corpus. A meaningful portion of the tweets is not unique to a single subtask's dataset.

*   **Pairwise Overlap Details:**

    *   **Call2Action vs. Violence:**
        *   Share **100** descriptions.
        *   These 100 descriptions represent **9.53%** of Call2Action's unique descriptions and **9.52%** of Violence's unique descriptions.
        *   Of the combined unique descriptions between *only* Call2Action and Violence (1999 descriptions), the 100 shared descriptions account for **5.00%**.
        *   -> A similar proportion of each dataset's content is shared with the other. The combined percentage suggests that while there's a direct overlap, the two datasets still largely contain distinct items from each other.

    *   **Call2Action vs. DBO:**
        *   Share **97** descriptions.
        *   These 97 descriptions represent **9.25%** of Call2Action's unique descriptions and **9.25%** of DBO's unique descriptions.
        *   Of the combined unique descriptions between *only* Call2Action and DBO (2001 descriptions), the 97 shared descriptions account for **4.85%**.
        *   -> Similar to the C2A-Violence pair, the overlap is proportionally balanced. The combined percentage again points to considerable distinct content between the two.

    *   **Violence vs. DBO:**
        *   Share **99** descriptions.
        *   These 99 descriptions represent **9.43%** of Violence's unique descriptions and **9.44%** of DBO's unique descriptions.
        *   Of the combined unique descriptions between *only* Violence and DBO (2000 descriptions), the 99 shared descriptions account for **4.95%**.
        *   -> Consistent pattern of balanced proportional overlap.

<u><p>Summary & Implications</p></u>

*   *Summary of Observations:*
    *   There's a consistent level of pairwise overlap (around 9-10% of each dataset involved in a pair is shared with the other).
    *   The "combined percentage" for each pair (around 5%) highlights that even when two datasets are considered together, the majority of their combined unique content is not shared.
    *   The overall 9.21% overlap across the entire corpus suggests that while subtasks might target different topics, some tweet instances are relevant to the annotation criteria of multiple subtasks.

*   *Implications*
    *   **Multi-Task Learning Potential:** The dataset overlap suggests that a **Multi-Task-Learning** ***(MTL)*** approach, where a single model predicts labels for multiple subtasks simultaneously, could be effective. This makes intuitive sense, as the overlap implies that the subtasks are not entirely distinct and likely share underlying characteristics. For instance, a tweet urging violent action to overthrow the German government could qualify as both a *'Call2Action'* and an *'Attack on DBO'*, indicating shared semantic features a single model could learn.
    *   **Data Leakage:** If datasets were to be merged into one dataset and then split into train/validation/test sets without considering these overlaps, identical tweets could contaminate the validation and test sets. This could lead to overly optimistic evaluation metrics for models that are effectively "seeing" test and validation data during training. In the case of merging the datasets, the dataset splits should be deduplicated to avoid the same text falling into more than one split (train, validation, or test).

##### <b>Analysing the overlapping tweets</b>

```python
df_overlapping_tweets = pd.DataFrame(list(overlaps["total"]), columns=["description"])

console.rule("[bold cyan]Overlapping Tweet Descriptions[/bold cyan]")
console.print(f"Total unique overlapping descriptions: {len(df_overlapping_tweets)}")

# Displays the first 3, last 3, and 3 random samples.
console.print("\n[bold]Head (first 5):[/bold]")
for tweet in df_overlapping_tweets["description"].head(5):
    console.print(f"- \"{tweet}\"")

console.print("\n[bold]Tail (last 5):[/bold]")
for tweet in df_overlapping_tweets["description"].tail(5):
    console.print(f"- \"{tweet}\"")

console.print("\n[bold]Random sample (5):[/bold]")
sample_size = min(5, len(df_overlapping_tweets))
for tweet in df_overlapping_tweets["description"].sample(sample_size, random_state=42):
    console.print(f"- \"{tweet}\"")
```

<u><p>Interpretation:</p></u>

We find, that the overlapping tweets consist of largely **substantive and meaningful tweets**, not merely collections of stopwords or noise.

*   **Observation:** The type of overlapping content ranges from short exclamations (e.g., `"Ihr seid soooooo böse"`, `"ja,und nun??"`) to longer statements (e.g., `"schau Dir die Kinder an. Die werden benutzt! Die sollte man rausholen, aber die Eltern dort lassen"`) and URLs.

*   **Key Implication:** ***Potential Data Leakage!*** Since these are actual, distinct tweet messages appearing in multiple datasets, their unmanaged presence across training, validation, and testing splits (if splits are done independently per subtask) could lead to models being evaluated on data they have effectively seen during training. This would compromise the integrity and meaningfulness of our performance metrics in the end.

*   **Conclusion:** Careful *filtering* and *data splitting* will likely be necessary to prevent our metrics from being clouded by tweets overlapping the datasets. (In an MTL setting)
