<p><b>C2A Per Class Frequency Distribution of the individual words (unfiltered)</b></p>

```python
# Gets the class corpuses
c2a_class_corpuses = data_utils.get_class_corpuses(df_call2action, "C2A")

# Tokenizes the class corpuses
c2a_true_tokens, c2a_true_sentences = data_utils.prepare_text_for_frequency_distribution_analysis(
    c2a_class_corpuses["C2A_true"]
)
c2a_false_tokens, c2a_false_sentences = data_utils.prepare_text_for_frequency_distribution_analysis(
    c2a_class_corpuses["C2A_false"]
)

# Creates the frequency distribution classes
c2a_true_freq_dist = FreqDist(c2a_true_tokens)
c2a_false_freq_dist = FreqDist(c2a_false_tokens)

# Creates a dictionary with the frequency distribution of the classes
freq_dist_dict_unfiltered = {"C2A_true": c2a_true_freq_dist, "C2A_false": c2a_false_freq_dist}

# Plots the frequency distributions as a bar plot
c2a_freq_dist_fig, c2a_freq_dist_ax = data_utils.create_frequency_distribution_plot(
    freq_dist_dict_unfiltered,
    title="C2A Frequency Word Distribution per Class (unfiltered)", 
    n_words=30,
    figsize=(16, 10),
    column_names=["Word", "Frequency"],
    show_grid=True
)
```
