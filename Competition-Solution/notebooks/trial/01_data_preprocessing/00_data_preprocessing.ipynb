### Data Preprocessing

The notebooks within the **`01_data_preprocessing`** folder detail the specific preprocessing steps applied to the trial dataset for each of the three subtasks:

-   **Subtask 1:** Call2Action
-   **Subtask 2:** Attacks on the Democratic Basic Order (FDGO)
-   **Subtask 3:** Violence Detection

Data preprocessing involves the cleaning and transforming of the raw text data into a suitable format. The preprocessing stage draws from the insights gained from the prior Data Profiling phase (documented in `00_data_profiling`).

The primary goals of the preprocessing steps outlined here are to:
*   Remove or replace noise and irrelevant elements commonly found in social media text that are not beneficial for transformer based models (e.g., URLs, special characters, excessive whitespace).
*   Handle dataset-specific artifacts (e.g., anonymization tokens like `[@GRP]`).

The specific preprocessing decisions and their implementation are guided by the characteristics identified during data profiling. A consistent and well-reasoned preprocessing pipeline ensures that the models learn from meaningful signals rather than noise. The cleaned data resulting from this stage will form the input for subsequent tokenization and model training.

### Train/Test/Validation Splitting

##### Converting the cleaned pandas dataframes into HuggingFace Datasets

```python
# Dataset Info objects
c2a_info = DatasetInfo(
    description="Binary detection of call-to-actions in German tweets from right-wing extremist network (2014-2016).",
    features=Features({
        "id": Value("string"),
        "description": Value("string"),
        "C2A": ClassLabel(names=["False", "True"])
    }),
    homepage="https://www.codabench.org/competitions/4963/",
    citation="GermEval 2025: A Shared Task for the Detection of Harmful Content on Social Media",
    license="Research use only - shared within GermEval 2025 competition",
    supervised_keys=("description", "C2A")
)

dbo_info = DatasetInfo(
    description="Four-class classification of statements against the free democratic basic order in German tweets.",
    features=Features({
        "id": Value("string"),
        "description": Value("string"),
        "DBO": ClassLabel(names=["nothing", "criticism", "agitation", "subversive"])
    }),
    homepage="https://www.codabench.org/competitions/4963/",
    citation="GermEval 2025: A Shared Task for the Detection of Harmful Content on Social Media",
    license="Research use only - shared within GermEval 2025 competition",
    supervised_keys=("description", "DBO")
)

vio_info = DatasetInfo(
    description="Binary detection of violence-related statements in German tweets from right-wing extremist network.",
    features=Features({
        "id": Value("string"),
        "description": Value("string"),
        "VIO": ClassLabel(names=["False", "True"])
    }),
    homepage="https://www.codabench.org/competitions/4963/",
    citation="GermEval 2025: A Shared Task for the Detection of Harmful Content on Social Media",
    license="Research use only - shared within GermEval 2025 competition",
    supervised_keys=("description", "VIO")
)
```

```python
# Converts the cleaned pandas dataframes into HuggingFace Datasets
c2a_dataset = Dataset.from_pandas(
    df=df_call2action_cleaned,
    features=c2a_info.features,
    info=c2a_info,
    preserve_index=False
)

dbo_dataset = Dataset.from_pandas(
    df=df_dbo_cleaned,
    features=dbo_info.features,
    info=dbo_info,
    preserve_index=False
)

vio_dataset = Dataset.from_pandas(
    df=df_violence_cleaned,
    features=vio_info.features,
    info=vio_info,
    preserve_index=False
)
```

```python
# Sequential Splitting (1. Split into train and test, 2. Split test into validation and test (80/10/10))
# Note: Shuffles the datasets during splitting to ensure that the data is randomly distributed

# c2a Splitting
c2a_splits = c2a_dataset.train_test_split(test_size=0.2, seed=42, shuffle=True)
c2a_temp_test = c2a_splits["test"].train_test_split(test_size=0.5, seed=42, shuffle=True)
c2a_dataset = DatasetDict({
    "train": c2a_splits["train"],
    "test": c2a_temp_test["train"],
    "validation": c2a_temp_test["test"]
})

# dbo Splitting
dbo_splits = dbo_dataset.train_test_split(test_size=0.2, seed=42, shuffle=True)
dbo_temp_test = dbo_splits["test"].train_test_split(test_size=0.5, seed=42, shuffle=True)
dbo_dataset = DatasetDict({
    "train": dbo_splits["train"],
    "test": dbo_temp_test["train"],
    "validation": dbo_temp_test["test"]
})

# vio Splitting
vio_splits = vio_dataset.train_test_split(test_size=0.2, seed=42, shuffle=True)
vio_temp_test = vio_splits["test"].train_test_split(test_size=0.5, seed=42, shuffle=True)
vio_dataset = DatasetDict({
    "train": vio_splits["train"],
    "test": vio_temp_test["train"],
    "validation": vio_temp_test["test"]
})
```

```python
# Saves the datasets to disk
c2a_dataset.save_to_disk("../../../data/processed/c2a/c2a_hf_dataset")
dbo_dataset.save_to_disk("../../../data/processed/dbo/dbo_hf_dataset")
vio_dataset.save_to_disk("../../../data/processed/vio/vio_hf_dataset")
```
