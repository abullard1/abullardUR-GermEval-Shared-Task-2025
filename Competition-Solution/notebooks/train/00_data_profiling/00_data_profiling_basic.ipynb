##### <b>Overlap between the subtask datasets</b>

```python
datasets_to_compare = [
    ("Call2Action", df_call2action), 
    ("Violence", df_violence), 
    ("DBO", df_dbo)
]

# Calculates the raw overlaps
overlaps = data_utils.calculate_dataset_overlaps(
    datasets=datasets_to_compare,
    column="description",
    show_details=False
)

# Displays metrics
percentage_results = data_utils.display_dataset_overlap_percentages(
    overlaps=overlaps,
    datasets=datasets_to_compare,
    column="description",
    show_pairwise=True
)
```

<u><p>Interpretation:</p></u>

The analysis of overlapping tweet descriptions across the three subtask datasets (Call2Action, Violence, and DBO) reveals the following:

*   **Overall Corpus Overlap:**
    *   Out of a total of **11,870 unique descriptions** across all three datasets combined, **7,447 descriptions (62.74%)** appear in *at least two* of these datasets.
    *   *Observation:* This indicates a **significantly higher level of shared content** compared to the trial dataset (9.21% overlap). The training dataset shows load of overlap, suggesting that many tweets are relevant to multiple subtasks at the same time.

*   **Pairwise Overlap Details:**

    *   **Call2Action vs. Violence:**
        *   Share **4,562** common descriptions.
        *   These 4,562 descriptions represent **67.25%** of Call2Action's unique descriptions and **59.04%** of Violence's unique descriptions.
        *   *Observation:* Strong overlap indicating that many tweets calling for action also contain violence-related content.

    *   **Call2Action vs. DBO:**
        *   Share **3,689** common descriptions.
        *   These 3,689 descriptions represent **54.37%** of Call2Action's unique descriptions and **50.51%** of DBO's unique descriptions.
        *   *Observation:* Substantial overlap showing that calls to action frequently involve attacks on democratic institutions.

    *   **Violence vs. DBO:**
        *   Share **4,194** common descriptions.
        *   These 4,194 descriptions represent **54.29%** of Violence's unique descriptions and **57.42%** of DBO's unique descriptions.
        *   *Observation:* High overlap indicating that violence-related content often targets democratic institutions.

<u><p>Summary & Implications</p></u>

*   *Summary of Observations:*
    *   The training dataset exhibits **dramatically higher overlap** (62.74%) compared to the trial dataset (9.21%), with more than half of each dataset's content being shared with other subtasks.
    *   All pairwise overlaps exceed 50%, indicating strong semantic relationships between the subtasks.
    *   The overlap pattern suggests that harmful content often exhibits multiple characteristics simultaneously.

*   *Implications*
    *   **Multi-Task Learning Potential:** The overlap strongly supports the multi task learning approach that we outlined during the trial dataset analysis and reinforces the fact that all subtasks share an intrinsic relationship with one another.

<u><p>Interpretation:</p></u>

The overlapping tweets in the training dataset consist of substantive and meaningful content. This has not changed from the trial dataset.

*   **Scale:** With **7,447 overlapping descriptions** (compared to 264 in the trial dataset), the training set presents a significant challenge for data leakage prevention and will ultimately result in less learning signals for an MTL model than initially anticipated or hoped for - after cleaning.

*   **Variable Lengths:** In the case of the training dataset, it becomes apparent, that the tweets have very differing lengths. They range from being quite short to being longer snippets of text, which will likely have implications on training and performance.
