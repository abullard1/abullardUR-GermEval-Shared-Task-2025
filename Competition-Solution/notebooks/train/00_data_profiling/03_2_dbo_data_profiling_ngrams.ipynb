```python
# Gets the class corpuses
dbo_class_corpuses = data_utils.get_class_corpuses(df_dbo, "DBO")

# Tokenizes the class corpuses
dbo_nothing_tokens, dbo_nothing_sentences = data_utils.prepare_text_for_frequency_distribution_analysis(
    dbo_class_corpuses["DBO_nothing"]
)
dbo_criticism_tokens, dbo_criticism_sentences = data_utils.prepare_text_for_frequency_distribution_analysis(
    dbo_class_corpuses["DBO_criticism"]
)
dbo_agitation_tokens, dbo_agitation_sentences = data_utils.prepare_text_for_frequency_distribution_analysis(
    dbo_class_corpuses["DBO_agitation"]
)
dbo_subversive_tokens, dbo_subversive_sentences = data_utils.prepare_text_for_frequency_distribution_analysis(
    dbo_class_corpuses["DBO_subversive"]
)

# Creates the n-gram frequency distributions for each class
dbo_nothing_bigram_freq = data_utils.generate_ngrams_from_tokens(dbo_nothing_tokens, n=2)
dbo_criticism_bigram_freq = data_utils.generate_ngrams_from_tokens(dbo_criticism_tokens, n=2)
dbo_agitation_bigram_freq = data_utils.generate_ngrams_from_tokens(dbo_agitation_tokens, n=2)
dbo_subversive_bigram_freq = data_utils.generate_ngrams_from_tokens(dbo_subversive_tokens, n=2)
dbo_nothing_trigram_freq = data_utils.generate_ngrams_from_tokens(dbo_nothing_tokens, n=3)
dbo_criticism_trigram_freq = data_utils.generate_ngrams_from_tokens(dbo_criticism_tokens, n=3)
dbo_agitation_trigram_freq = data_utils.generate_ngrams_from_tokens(dbo_agitation_tokens, n=3)
dbo_subversive_trigram_freq = data_utils.generate_ngrams_from_tokens(dbo_subversive_tokens, n=3)

# Creates a dictionary with the frequency distributions
ngram_freq_dict_unfiltered = {
    "DBO_nothing_bigram": dbo_nothing_bigram_freq, 
    "DBO_criticism_bigram": dbo_criticism_bigram_freq,
    "DBO_agitation_bigram": dbo_agitation_bigram_freq,
    "DBO_subversive_bigram": dbo_subversive_bigram_freq,
    "DBO_nothing_trigram": dbo_nothing_trigram_freq, 
    "DBO_criticism_trigram": dbo_criticism_trigram_freq,
    "DBO_agitation_trigram": dbo_agitation_trigram_freq,
    "DBO_subversive_trigram": dbo_subversive_trigram_freq
}

# Plots the n-gram distributions as a bar plot
dbo_bigram_fig, dbo_bigram_ax = data_utils.create_ngram_frequency_plot(
    ngram_freq_dict_unfiltered,
    title="DBO n-gram Distribution per Class (unfiltered)", 
    n_items=20,
    figsize=(20, 24),
    show_grid=True
)
```
