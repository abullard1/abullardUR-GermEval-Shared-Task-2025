<p><b>VIO n-gram Frequency Distribution (unfiltered)</b></p>
<p><i><li>Bar Plot</li><i><p>

```python
# Gets the class corpuses
vio_class_corpuses = data_utils.get_class_corpuses(df_violence, "VIO")

# Tokenizes the class corpuses
vio_true_tokens, vio_true_sentences = data_utils.prepare_text_for_frequency_distribution_analysis(
    vio_class_corpuses["VIO_true"]
)
vio_false_tokens, vio_false_sentences = data_utils.prepare_text_for_frequency_distribution_analysis(
    vio_class_corpuses["VIO_false"]
)

# Creates the n-gram frequency distributions for each class
vio_true_bigram_freq = data_utils.generate_ngrams_from_tokens(vio_true_tokens, n=2)
vio_false_bigram_freq = data_utils.generate_ngrams_from_tokens(vio_false_tokens, n=2)
vio_true_trigram_freq = data_utils.generate_ngrams_from_tokens(vio_true_tokens, n=3)
vio_false_trigram_freq = data_utils.generate_ngrams_from_tokens(vio_false_tokens, n=3)

# Creates a dictionary with the frequency distributions
ngram_freq_dict_unfiltered = {"VIO_true_bigram": vio_true_bigram_freq, "VIO_false_bigram": vio_false_bigram_freq, 
                              "VIO_true_trigram": vio_true_trigram_freq, "VIO_false_trigram": vio_false_trigram_freq}

# Plots the n-gram distributions as a bar plot
vio_bigram_fig, vio_bigram_ax = data_utils.create_ngram_frequency_plot(
    ngram_freq_dict_unfiltered,
    title="VIO n-gram Distribution per Class (unfiltered)", 
    n_items=20,
    figsize=(16, 10),
    show_grid=True
)
```
